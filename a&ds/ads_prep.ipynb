```python
# DFS ARE STACKS
"""
Depth-First Search aims to go as deep as possible along a single path before it ever looks at a "sibling" node.
Because a stack is LIFO (Last-In, First-Out), the most recently discovered node is the one that gets popped and processed next. 
This forces the algorithm to follow the newest lead immediately, driving it deeper into the graph.
"""

def dfs_recursive(graph, node, visited = None):
    # Initialised the visited set
    if visited is None:
        return set()
    
    # Mark node as seen
    visited.add(node)
    print(node, end = "")
    
    # Recur
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs_recursive(graph, node, visited)
```

```python
# BFS ARE QUEUES
"""
Breadth-First Search aims to explore a graph in layers. It visits all nodes at distance 1, then all nodes at distance 2, and so on.
Because a queue is FIFO (First-In, First-Out), the neighbors discovered first are processed first. 
This ensures that the algorithm finishes the current "layer" before it ever starts moving deeper into the next layer.
"""
def bfs_iterative(graph, start_node):
    # Initialise 
    visited = set()
    queue = [start_node]
    visited.add(start_node)
    
    while len(queue)>0:
        # Remove the first element which is the oldest
        current_node = queue.pop(0)
        print(current_node, end="")
        
        for neighbor in graph[current_node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```
