```python
# Factorial 
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
print(factorial(3))
```

```python
# Summing a list of numbers
def summation(n):
    if n <= 0:
        return 0
    return n + summation(n-1)
```

```python
# x^n
def power(x,n):
    if n <= 0:
        return 0
    return x * power(x, n-1)
```

```python
# Sum of list elements
def sum_lists(numbers):
    if len(numbers) == 0:
        return 0
    return numbers[0] + sum_lists(numbers[1:])

print(sum_lists([1,2,3]))
```

```python
# Find maximum number
def find_max(numbers):
    if len(numbers) == 1:
        return numbers[0]
    return max(numbers[0], find_max(numbers[1:]))
print(find_max([1,2,3]))
    
```

```python
# Nested structures
# Calculate the sum of all numbers in a list that may contain other lists
def sum_nested(data):
    total = 0
    
    for item in data:
        if isinstance(item, list):
            # If item is a list, recursively sum
            total += sum_nested(item)
        else:
            total += item
    return total

nested = [1, [2, 3], [4, [5, 6]], 7]
```

```python
# Same thing but it counts instead
def count_nested(data):
    count = 0
    
    for item  in data:
        if isinstance(item, list):
            # Item is a list, recursively run
            count += count_nested(item)
        else:
            count += 1
    return count
```

```python
"""
Linear Search to find if an element exists in a list
"""
def linear_search(numbers, target):
    # emptyt list
    if len(numbers) == 0:
        return False
    
    if numbers[0] == target:
        return True
    
    # search rest of list
    return linear_search(numbers[1:], target)

```

```python
"""
Binary Search
These first need to be ordered
which is done via list.sort()
"""

def binary_search(numbers, target, left=0, right=None):
    # If no right is given, set it to the last index of the list
    if right is None:
        right = len(numbers) - 1
        
    # Nothing left to search
    if left > right :
        return False
    
    # Get middle index
    middle = (left + right) // 2

    if numbers[middle] == target:
        return True
    
    
    if target < numbers[middle]:
        # If the target is smaller than the binary tree
        return binary_search(numbers, target, left, middle - 1)
    else:
        return binary_search(numbers, target, middle + 1, right)
```
