```python
"""Recursive"""
def bfs_connected(graph):
    visited = set()
    
    def bfs(queue):
        if not queue:
            return
        node = queue.pop(0)
        for j in range(len(graph[node])):
            if graph[node][j] == 1 and j not in visited:
                visited.add(j)
                queue.append(j)
        bfs(queue)
    visited.add(0)
    bfs([0])
    return len(visited) == len(graph)
```


Graph traversal algorithms can be used to determine whether a graph contains any cycles. Given a directed graph, expressed as adjacency list, write a recursive depth-first search algorithm that determines whether the graph contains any cycles. Use the following function heading:

```python
def dfs_contains_cycles(graph): pass
```

where graph is a dict[obj, set[obj]], and the function returns a boolean. You are allowed to use additional functions (probably makes it easier).

```python
"""Recursive"""
def dfs_contains_cycles(graph):
    # Visited set to track visited ones, current path to keep track of the path taken
    visited = set()
    current_path = set()
    
    def dfs(graph, node, visited, current_path):
        # If node is in current path, the cycle exists
        if node in current_path:
            return True
        
        # Add the node to both sets if it isn't in either
        current_path.add(node)
        visited.add(node)
        
        # Recurse
        for neighbor in graph[node]:
            if neighbor not in visited:
                # If a cycle is found, return True
                if dfs(graph, neighbor, visited, current_path):
                    return True
        # If no cycle is found, backtrack to prevent "false" loops from being made
        # E.g DFS goes 0 → 1 → 2, so current_path = {0, 1, 2}. Then it backtracks to explore 0 → 3 → 1. 
        # If you don't remove then it'll think there is a cycle however there isn't.
        current_path.remove(node)
        # Return False since nothing was found
        return False
    
    # Ensure every node is visited
    # If you only started DFS from node 0, you'd never visit nodes 3 and 4! So the outer loop ensures you start a DFS from every unvisited node, guaranteeing the whole graph is checked.
    for node in graph:
        if node not in visited:
            if dfs(graph, node, visited, current_path):
                return True
    return False

graph = {
  'a': {'b', 'c'},
  'b': {}, 
  'c': {} 
} # no cycles
print(dfs_contains_cycles(graph))
```

```python
def dfs_recursive(graph):
    visited = set()
    current_path = set()
    
    def dfs(graph, node, visited, current_path):
        if node in current_path: return True
        visited.add(node)
        current_path.add(node)
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                if dfs(graph, node, visited, current_path):
                    return True
        current_path.remove(node)
        return False
    for node in graph:
        if node not in visited:
            if dfs(graph, node, visited, current_path):
                return True
    return False
```

```python
"""Iterative"""
def DFS_is_acyclic(graph) -> bool:
    """
    Checking if the DFS is acylic
    """	
    # Base case
    if not graph:
        return True
    
    # Init of sets
    visited = set()
    n = len(graph)
    in_progress = set()

    # Starting DFS from node 0 
    stack = []

    for node in graph:
        # For every node in the graph, if it wasn't been visited, add to unvisited stack
        if node not in visited:
            stack.append((node, False))

    # While the stack exists
    while stack:
        node, is_leaving = stack.pop()

        # If True, exiting since all neighbours are done
        # If False, entering node for the first time
        if is_leaving:
            in_progress.remove(node)
            visited.add(node)
        else:
            if node in in_progress: # Return a false because it shouldn't be
                return False
            in_progress.add(node)
            stack.append((node, True)) # To return and exit later
            for neighbor in graph[node]:
                if neighbor not in visited:
                    stack.append((neighbor, False))
    return True
```

The m-colouring problem asks how to colour the vertices of a graph, using m different colours, in such a way that every neighbouring vertices have different colours.

Write a recursive function, using backtracking, that for an undirected connected graph returns a colouring assignment. (You are allowed write helper functions) 
The function is provided the graph as an adjacency list, a dictionary of lists, and m a list of colours. The function should return a dictionary with the vertices as items colours as values, or an empty dictionary if no assignment exists.


```python
def m_colouring(graph, m):
    # Created assignment map of colours which have already been assigned
    assignment = {}
    
    def backtrack(node):
        # If everything is coloured then return the assignment
        if len(assignment) == len(graph):
            return assignment
            
        # Try each colour one by one
        for colour in m:
            valid = True
            # Checking the nodes neighbors and make sure none of them have this colour
            for neighbour in graph[node]:
                if assignment.get(neighbour) == colour:
                    valid = False
                    break
            # If no neighbour have this colour, assign colour to this node
            if valid:
                assignment[node] = colour
                # Recursively colour the next node
                result = backtrack(node + 1)
                # If the next node is coloured, keep recursing until final node
                if result:
                    return result
                del assignment[node]
        # Nothing has worked hence return False
        return False
    
    result = backtrack(0)
    return result if result else {}

graph = { 0: [1, 2], 1 : [0], 2: [2]}
m = ["blue", "red"] 
print(m_colouring(graph, m))
```

```python
def m_colouring(graph, m):
    assignment = {}
    
    def backtrack(node):
        if len(assignment) == len(graph):
            return True
        
        for colour in m:
            valid = True
            for neighbor in graph[node]:
                if assignment(neighbor) == colour:
                    valid = False
                    break
            if valid:
                assignment[node] = colour
                result = backtrack(node + 1)
                if result: return result
                del assignment[node]
        return False
    
    result = backtrack(0)
    return result if result else {}
```
